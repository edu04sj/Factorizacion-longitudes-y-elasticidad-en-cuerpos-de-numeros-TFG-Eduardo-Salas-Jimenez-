```sage
# Definir d (vos lo cambiás)
d = 21   # ejemplo: Q(sqrt(-5))

# Definir el cuerpo cuadrático imaginario
K.<a> = QuadraticField(-d)

R.<x> = QQ[]
R.<x> = K[]   # clave

# Anillo de enteros
OK = K.ring_of_integers()

# Grupo de clases
Cl = OK.class_group()

# Número de clase
h = Cl.order()

A = Cl.gens()[0].ideal()
B = Cl.gens()[1].ideal()

# Imprimir resultados
print("Cuerpo K =", K)
print("Anillo de enteros OK =", OK)
print("Grupo de clases =", Cl)
print("Número de clase =", h)
show(A.gens_reduced())
show(B)
```

```sage
I = OK.ideal([5, a + 2])
J = OK.ideal([2, a + 1])

show((I^2).gens_reduced())
show((J^2).gens_reduced())
```
