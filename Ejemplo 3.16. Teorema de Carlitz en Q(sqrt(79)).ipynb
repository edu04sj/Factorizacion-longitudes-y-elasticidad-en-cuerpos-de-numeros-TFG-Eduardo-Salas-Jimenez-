```sage
def prime_rep_search(c, K, Cl, n):

    # Buscamos ideales primos en la clase de ideales c.
    el_primo = []
    m = 1

    while len(el_primo) < n:

        # Factorizamos los ideales generados por primos racionales.
        for p in prime_range(m, m + 1000):
            for P, e in factor(K.ideal(p)):
                if Cl(P) == c:
                    el_primo.append(P)

                    if len(el_primo) == n:
                        break

            if len(el_primo) == n:
                break

        m += 1000

    return el_primo
```

```sage
# Definimos el cuerpo cuadrático K.
K.<a> = QuadraticField(79)

# Calculamos el anillo de enteros y el grupo de clases de K.
OK = K.ring_of_integers()
Cl = OK.class_group()
h = OK.class_number()
print("Cl(K) = ", Cl)

# Elegimos las clases de ideales según la estructura del grupo de clases.
if h <= 2:
    I1 = OK.ideal(0)
    I2 = OK.ideal(0)
    I3 = OK.ideal(0)
    show("OK es semi-factorial")

elif len(Cl.gens()) == 1:
    g = Cl.gens()[0]
    I1 = g
    I2 = g
    I3 = g^(-2)

else:
    g1 = Cl.gens()[0]
    g2 = Cl.gens()[1]
    I1 = g1
    I2 = g2
    I3 = (g1 * g2)^(-1)

# Mostramos las clases de ideales obtenidas.
show(LatexExpr(r"I_1 ="), I1)
show(LatexExpr(r"I_2 ="), I2)
show(LatexExpr(r"I_3 ="), I3)
```

```sage
# Buscamos ideales primos que representan las clases de I_1, I_2 e I_3.
P1 = prime_rep_search(I1, K, Cl, 1)[0]
show(LatexExpr(r"P_1 ="), P1)

P2 = prime_rep_search(I2, K, Cl, 1)[0]
show(LatexExpr(r"P_2 ="), P2)

P3 = prime_rep_search(I3, K, Cl, 1)[0]
show(LatexExpr(r"P_3 ="), P3)
```

```sage
# Buscamos ideales primos que representan las clases de I_1^{-1}, I_2^{-1} e I_3^{-1}.
Q1 = prime_rep_search(I1^-1, K, Cl, 1)[0]
show(LatexExpr(r"Q_1 ="), Q1)

Q2 = prime_rep_search(I2^-1, K, Cl, 1)[0]
show(LatexExpr(r"Q_2 ="), Q2)

Q3 = prime_rep_search(I3^-1, K, Cl, 1)[0]
show(LatexExpr(r"Q_3 ="), Q3)
```

```sage
# Calculamos los productos de ideales que definen eta_1 y eta_2.
n1 = P1 * P2 * P3
show(LatexExpr(r"\eta_1 ="), n1)

n2 = Q1 * Q2 * Q3
show(LatexExpr(r"\eta_2 ="), n2)

# Calculamos los productos de ideales que definen omega_1, omega_2 y omega_3.
w1 = P1 * Q1
show(LatexExpr(r"\omega_1 ="), w1)

w2 = P2 * Q2
show(LatexExpr(r"\omega_2 ="), w2)

w3 = P3 * Q3
show(LatexExpr(r"\omega_3 ="), w3)

# Verificamos la igualdad entre ambos productos.
show(LatexExpr(r"\eta_1\eta_2 ="), n1 * n2)
show(LatexExpr(r"\omega_1\omega_2\omega_3 ="), w1 * w2 * w3)
```
