```sage
# Definir d (vos lo cambiás)
d = 23   # ejemplo: Q(sqrt(-5))

# Definir el cuerpo cuadrático imaginario
K.<a> = QuadraticField(-d)

R.<x> = QQ[]
R.<x> = K[]   # clave

# Anillo de enteros
OK = K.ring_of_integers()

# Grupo de clases
Cl = OK.class_group()

# Número de clase
h = Cl.order()

A = Cl.gens()[0].ideal()

# Imprimir resultados
print("Cuerpo K =", K)
print("Anillo de enteros OK =", OK)
print("Grupo de clases =", Cl)
print("Número de clase =", h)
show(A)
```

```sage
I = OK.ideal(A.gens_reduced())

show((I^3).gens_reduced())
```
